```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('./Datasets/5-sales_data_sample.csv', encoding='unicode_escape')
df.head()
```

```python
df.dropna(inplace = True)
```

```python
x = df[['SALES', 'QUANTITYORDERED', 'PRICEEACH']]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(x)
```

```python
from sklearn.cluster import KMeans
inertia = []
K = range(1,10)
for k in K:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(X_scaled)
    inertia.append(model.inertia_)
```

```python
import seaborn as sns
sns.lineplot(x = K, y = inertia)
```

```python
optimal_k = 3
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)
df[['SALES', 'QUANTITYORDERED', 'PRICEEACH']].head()
```

```python
df.groupby('Cluster')[['SALES', 'QUANTITYORDERED', 'PRICEEACH']].mean()
```
