```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN
```

```python
# Step 2: Load Dataset
# Make sure google.csv is in the same folder as notebook

df = pd.read_csv("./Datasets/google.csv")

df.head()
df.columns()
```

```python
# Step 3: Take Open Column
# Convert Open column into numeric values

data = pd.to_numeric(df['Open'], errors='coerce')

# Remove missing values
data = data.dropna()

# Convert into numpy array
data = data.values.reshape(-1, 1)

print(data.shape)
```

```python
# Step 4: Normalize Data
scaler = MinMaxScaler(feature_range=(0,1))
data_scaled = scaler.fit_transform(data)
```

```python
# Step 5: Create Training Data
x_train = []
y_train = []

# Use previous 60 days to predict next day
for i in range(60, len(data_scaled)):
    x_train.append(data_scaled[i-60:i, 0])
    y_train.append(data_scaled[i, 0])

# Convert into numpy arrays
x_train = np.array(x_train)
y_train = np.array(y_train)

print(x_train.shape)
print(y_train.shape)
```

```python
# Step 6: Reshape for RNN
# Shape required: (samples, time_steps, features)

x_train = np.reshape(x_train, (x_train.shape[0], x_train.shape[1], 1))

print(x_train.shape)
```

```python
# Step 7: Build RNN Model
model = Sequential()

model.add(SimpleRNN(50, return_sequences=True, input_shape=(60,1)))
model.add(SimpleRNN(50))
model.add(Dense(1))

model.compile(optimizer='adam', loss='mean_squared_error')

model.summary()
```

```python
# Step 8: Train Model
history = model.fit(
    x_train,
    y_train,
    epochs=20,
    batch_size=32
)
```

```python
# Step 9: Prepare Test Data
x_test = []
real = data_scaled[60:]

for i in range(60, len(data_scaled)):
    x_test.append(data_scaled[i-60:i, 0])

x_test = np.array(x_test)

# Reshape for RNN
x_test = np.reshape(x_test, (x_test.shape[0], x_test.shape[1], 1))

print(x_test.shape)
```

```python
# Step 10: Predict Values
predicted = model.predict(x_test)

# Convert back to original values
predicted = scaler.inverse_transform(predicted)

# Real values in original scale
real = scaler.inverse_transform(real)
```

```python
# Step 11: Plot Results
plt.figure(figsize=(12,6))

plt.plot(real, color='red', label='Real Price')
plt.plot(predicted, color='blue', label='Predicted Price')

plt.title('Google Stock Price Prediction using RNN')
plt.xlabel('Time')
plt.ylabel('Stock Price')
plt.legend()
plt.show()
```
