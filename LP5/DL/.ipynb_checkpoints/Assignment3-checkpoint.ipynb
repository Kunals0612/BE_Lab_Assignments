```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Dropout, Flatten, Dense
```

```python
train_df = pd.read_csv('fashion-mnist_train.csv')
test_df  = pd.read_csv('fashion-mnist_test.csv')

class_names = ['Tshirt/Top','Trousers','Pullover','Dress','Coat',
               'Sandal','Shirt','Sneaker','Bag','Ankle Boot']

# Separate features and labels
x_train = train_df.iloc[:, 1:].to_numpy().reshape(-1, 28, 28, 1) / 255
y_train = train_df.iloc[:, 0].to_numpy()

x_test  = test_df.iloc[:, 1:].to_numpy().reshape(-1, 28, 28, 1) / 255
y_test  = test_df.iloc[:, 0].to_numpy()
train_df.head()
```

```python
plt.figure(figsize=(10, 10))
for i in range(25):
    plt.subplot(5, 5, i+1)
    plt.xticks([]); plt.yticks([]); plt.grid(False)
    plt.imshow(x_train[i], cmap=plt.cm.binary)
    plt.xlabel(class_names[y_train[i]])
plt.show()
```

```python
model = Sequential([
    Conv2D(64, (3,3), activation='relu', input_shape=(28,28,1)),  # detect features
    MaxPooling2D((2,2)),                                           # shrink feature map
    Dropout(0.3),                                                  # prevent overfitting
    Flatten(),                                                     # 2D → 1D
    Dense(32, activation='relu'),
    Dense(10, activation='sigmoid')                                # 10 classes output
])

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=50, batch_size=1200, validation_split=0.05)
```

```python
# Accuracy score
evaluation = model.evaluate(x_test, y_test)
print(f"Accuracy: {evaluation[1]}")

# Get predicted classes
y_pred = model.predict(x_test).argmax(axis=-1)

# Plot predictions on test images
plt.figure(figsize=(10, 10))
for i in range(25):
    plt.subplot(5, 5, i+1)
    plt.xticks([]); plt.yticks([]); plt.grid(False)
    plt.imshow(x_test[i], cmap=plt.cm.binary)
    plt.title(class_names[y_pred[i]])
plt.show()
```
